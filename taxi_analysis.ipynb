```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("Taxi_Trips_-_2022_20260216.csv")
df.head()
```

```python
print(df.columns)
```

```python
distance_col = "Trip Miles"
fare_col = "Fare"   # change to "Trip Total" if you want total amount instead
```

```python
distance_col = "Trip Miles"
fare_col = "Trip Total"
```

```python
df[distance_col] = pd.to_numeric(df[distance_col], errors="coerce")
df[fare_col] = pd.to_numeric(df[fare_col], errors="coerce")
```

```python
df_clean = df.dropna(subset=[distance_col, fare_col]).copy()

print("Original shape:", df.shape)
print("Cleaned shape:", df_clean.shape)
```

```python
zero_distance_with_fare = df_clean[
    (df_clean[distance_col] == 0) & (df_clean[fare_col] > 0)
]

print("Rows where distance = 0 but fare > 0:", len(zero_distance_with_fare))
zero_distance_with_fare[[distance_col, fare_col]].head(10)
```

```python
zero_fare_with_distance = df_clean[
    (df_clean[distance_col] > 0) & (df_clean[fare_col] == 0)
]

print("Rows where fare = 0 but distance > 0:", len(zero_fare_with_distance))
zero_fare_with_distance[[distance_col, fare_col]].head(10)
```

```python
print("Summary")
print("-------")
print("0 distance, but fare > 0:", len(zero_distance_with_fare))
print("distance > 0, but 0 fare:", len(zero_fare_with_distance))
```
